# AA Engineering Toolkit
Engineering analyses in Python, organised by discipline. Each notebook uses the `engtoolkit`
package, so the same calculations power these notebooks and the web app.

| Folder | Notebooks |
|---|---|
| `structural/` | statics (FBD, determinacy, equilibrium or FEM), bending, torsion & warping, shear-flexible beams, buckling, Maxwell-Mohr |
| `thermo/` | pipe sizing, pressure drop and insulation |
| `asme/` | ASME VIII Div 1 pressure vessels, B31.3 process piping |
| `electrical/` | battery modelling (coming in a later release) |

### The `engtoolkit` package
- `engtoolkit.materials` – shared solid and fluid properties (`get_material`, `fluid`)
- `engtoolkit.catalogs` – pipe schedules, IPE profiles, fittings and valve costs
- `engtoolkit.thermo` – fluid properties, steam tables, pipe flow, heat transfer, insulation, cycles
- `engtoolkit.structural` – sections, beams, determinacy, equilibrium and FEM solvers, torsion
- `engtoolkit.run("<tool id>", ...)` – run any registered tool, e.g. `"thermo.pipe_pressure_drop"`


How to set up the virtual python environment and install the toolkit:
```bash
# from the repository root
python -m venv venv
venv\Scripts\activate            # Windows (macOS/Linux: source venv/bin/activate)
pip install -e ".[notebook]"      # installs engtoolkit and the Jupyter kernel
```
Then pick the `venv` interpreter as the notebook kernel in VS Code.
