# Shear Flexible Beam (Timoshenko)

- When shear deformation matters: short, deep beams and sandwich/composite sections
- Shear correction factor $\kappa$ (5/6 for rectangles)
- Shear parameter $\phi = \frac{12EI}{\kappa G A L^2}$
- Compare Euler-Bernoulli vs Timoshenko with `FEMSolver(beam, theory="EB")` and `FEMSolver(beam, theory="TB")`
