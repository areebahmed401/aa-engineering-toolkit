# Overview of Structural Analysis and Design

1. Simplify model - Use assumptions to simplify the analysis (e.g., plane sections remain plane, small deformations, etc.)
2. Get MoI- Use formulas for different shapes
3. Check if they are movable - Counting criterion & pole plan
4. Statically determinable
   1. Calculate forces and moments using equilibrium equations
5. Statically indeterminable
   1. Calculate forces and moments using different methods
      1. Force method (suitable for hand calculations)
         1. Betti's theorem
         2. Maxwell's theorem
      2. slope-deflection method (Easy to program)
      3. Ritz method (Suitable for hand calculations)
      4. Finite element method (Direct stiffness method, )
6. Shear Flux calculation in cross section- Use formulas for different shapes
   1. Calculate the shear flow and shear stress distribution across the cross section
   2. Determine the maximum shear stress and compare it with the allowable shear stress for the material
   3. Design the cross section to ensure that it can safely carry the applied shear forces without failure
   4. Shear center calculation- Use formulas for different shapes
   5. Shear Flow distribution
7. Get, Moment, shear and deflection diagrams- Use the different end and loading conditions
8. Torsion analysis- Use formulas for different shapes
9. Check between shear rigid vs shear flexible beam (Timoshenko beam theory)
10. Plasticity behavior
11. Composite materials
12. Stability analysis (Columns)
    1.Energy Method (Ritz method - for all types of beams)
        1. Perfect vs imperfect systems
    2.Euler buckling (uniform cross section)
    3. Global vs local buckling
