# Warping and Torsion

The rectangular cross-section is warping-free if the condition
$$ \frac{B}{\left( G h \right)_{B}} = \frac{H}{\left( G h \right)_{H}} $$
applies.
The term $G h$ is denoted as shear stiffness of the skin!
$$ G = \frac{E}{2(1+\nu)} $$

### Saint-Venant Torsion
Warping free Torsion

Twist $\theta'$
 $$\theta'= \frac{T}{G J_T}$$
Integrate twist over the cross-section
 $$\theta(x) = \int_0^x \theta' dx$$
 

*Closed Thin Walled Cross-Section*

Bredt's first Formula
$$ M_t =2At_o$$
Bredt's second Formula
$$ GJ_T= \frac{4A^2}{\int (\frac{ds}{G h})}$$
Torsional Section Modulus
$$ W_T =2Ah_{min}$$

## Warping Torsion
Here there is stress due to torsion and warping. If you look at it from the top the warping is sort of like bending.
Therefore,total torsional moment is
$$ T= M_B + M_{SV}$$
$$ T= -\theta'''C_TE+ \theta'GJ_T$$

Given a cantilever beam under torque, see below.

![](../Images/Torsion%20%20on%20Cantilever.png "Example Warping")

The boundary conditions are:
- $\displaystyle{\quad\quad\vartheta(x_1=0) = 0}$
- $\displaystyle{\quad\quad \vartheta'(x_1=0) = 0}$
- $\displaystyle{\quad\quad \vartheta''(x_1=l) = 0}$

The solution of the differential equation and the derivatives are:

\begin{align}
M_{SV} & = & G I_T \vartheta' = \frac{G I_T \mu}{l \alpha^2} \left( \tanh \alpha \sinh \alpha \xi - \cosh \alpha \xi +1 \right)\\
    & = & T  \left( \tanh \alpha \sinh \alpha \xi - \cosh \alpha \xi +1 \right)\\
M_B    & = & - E C_T \vartheta''' = - \frac{E C_T \mu}{l^3} \left( \tanh \alpha \sinh \alpha \xi - \cosh \alpha \xi \right)\\
       & = & - T \left( \tanh \alpha \sinh \alpha \xi - \cosh \alpha \xi \right)
\end{align}
with
$$\xi= \frac{x_1}{l}\ ; \ \alpha^2=\frac{GI_Tl^2}{E C_T} \ ; \ \mu=\frac{Tl^3}{E C_T}$$

$$ \theta = A_1 +A_2\cosh(\alpha \xi) + A_3\sinh(\alpha \xi) +\frac{\mu}{\alpha^2}\xi$$

```python


mat=get_material("Aluminum")
print(mat.name)
s= I_beam(300/1000,10/1000,400/1000,10/1000)
b=beam(material=mat, length=1, shape=s)
print(b.get_mass())
t= Torsion(b, TorsionalMoment=200, position=1)

print("E =", b.material.E)
print("G =", b.material.G)
print("J_t =", b.shape.J_torsion)
print("L =", b.length)
print("C_t =", b.shape.C_torsion)

print("alpha =", t.alpha)
print("xi =", t.xi)
print("mu =", t.mu)
print("A1 =", t.A1)
print("A2 =", t.A2)
print("A3 =", t.A3)
theta = t.theta()
print("theta =", theta)
print("degrees =", rads_to_degrees(theta))
```
